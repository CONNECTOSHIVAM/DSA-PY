Insertion Sort:


```python
#Q. write a code for insertion sort.

##Code Analysis:

# Num of comparision in BC: (n - 1)
# Num of comparision in WC: n(n-1)/2

# Num of swaps in BC: 0
# Num of swaps in WC: n(n-1)/2

# Time Complexity BC : O(n)
# Time Complexity WC : O(n^2)
# Space Complexity : O(1) --- Inplace sorting Algorithm
# Insertion sort provide the Stable sorting


#Methood definations.
def isInsertionSort(a):
    n = len(a)


    for j in range(1,n):
        key = a[j]
        i = j- 1
        while(i>=0 and a[i]>key):
            a[i+1] = a[i]
            i -= 1  
        a[i+1] = key
    return a
            



#Driver code:
arr = [ 85,1,49,97,86,15]
print(f"Before sorting array was: {arr}")
result = isInsertionSort(arr)
print(f'After the soring array: {result}')
```

```python
#write a program to reverse the numbers.

def reverseX(n):

    a = abs(n)
    rev = 0

    while(a>0):
        b = a%10
        rev = rev*10+b
        a = a//10

    
    return rev
```
